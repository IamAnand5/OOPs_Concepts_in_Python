## Aggregation (Has-A relationship)

```python
# example of Aggregation
class Customer:
    
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address    #Here address is used by Cutomer class it is called Aggregation

        #attribute(address).attribute of address(city,pin,state)
    def print_address(self):
        print(self.address.get_city(), self.address.pin, self.address.state)
        
    def edit_profile(self,new_name,new_city,new_pin,new_state):
        self.name = new_name
        self.address.edit_address(new_city,new_pin,new_state)
        
class Address:
    def __init__(self,city,pin,state):
        self.__city = city    # if we make city as private 
        self.pin = pin
        self.state = state
        
    def get_city(self):      #we make getter method for variable city
        return self.__city

    
    def edit_address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state

#object of Address     
add1 = Address('Bilaspur',174001,"HP")

#object of Customer (Aggregation performed here)
cust = Customer('Anand','Male',add1)

cust.print_address()

cust.edit_profile('Ankit','Mumbai',110991,'Maharashtra')
cust.print_address()

# method example
# what about private attribute
```

###### In above if we make city as private (__city)variable in Address class. It is not possible to access the (__city) variable in Customer class. To do  so getter and setter methods.

```python
# Super Keyword-: By using 'Super' keyword we can call parent class items. It is a way to access parent's methods.

class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a phone")
        
class SmartPhone(Phone):
    def buy(self):
        print("Buying a Smartphone")
        # Syntax to call parent ka buy()
        super().buy()
        
s = SmartPhone(20000,"Apple",13)

s.buy()
```

```python
# super -> constructor


class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    
        
class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print("Inside smartphone constructor")
        # here calling parent's constructor from child's constructor using super
        super().__init__(self, price, brand, camera)
        self.os = os
        self.ram = ram
        print("Inside smartphone constructor")
        
s = SmartPhone(20000,"Samsung",12,"Android", 2)

print(s.os)
print(s.brand)
```

### Types of Inheritance
   1.) Single Inheritance
   2.) Multilevel Inheritance
   3.) Hierarchical Inheritance
   4.) Multiple Inheritance
   5.) Hybrid Inheritance

```python
# Example-: Single Inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print("Inside the Phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a Phone")
        
class SmartPhone(Phone):
    pass

SmartPhone(10000,"Apple","13px").buy()
```

```python
# Example: Multilevel Inheritance

class Product:
    def review(self):
        print("Product customer review")
        
class Phone(Product):
    def __init__(self, price, brand, camera):
        print("Inside Phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a Phone")
        
class SmartPhone(Phone):
    pass

s = SmartPhone(20000,"Apple",12)

s.buy()
s.review()
```

```python
# Example: Hierarchical Inheritance

class Phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a Phone")
        
class SmartPhone(Phone):
    pass

class FeaturePhone(Phone):
    pass

SmartPhone(10000,"Samsung","13px").buy()
FeaturePhone(12000,"Lava","12px").buy()
```

```python
# Example: Multiple Inheritance

class Phone:
    def __init__(self, price, brand, camera):
        print("Inside Phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera
        
    def buy(self):
        print("Buying a Phone")
        
class Product:
    def review(self):
        print("Customer review")
        
class SmartPhone(Phone, Product):
    pass

s = SmartPhone(20000,"Lava",12)

s.buy()
s.review()
```

### Abstraction-: means 'hidden'. Implementation details are hidden.
#### What is Abstract Class?
An abstract class is a class which has at least one abstract method. Although it may have more than one abstract() or concrete(). Two main things has to be done to be an abstract class.
    1.) Need to inherit the ABC class.
    2.) Need to have atleast one abstract method.
    
  We can not create object of Abstract class.

#### What is Abstract method?
An abstract method is a method which has no code inside its body. i.e. no implementation is done of that method. It has a decorator '@abstractmethod'

```python
# Abstraction

from abc import ABC, abstractmethod
class BankApp(ABC):
    
    def database(self):
        print('connected to database')
        
    @abstractmethod
    def security(self):
        pass
```

```python
class MobileApp(BankApp):
    
    def mobile_login(self):
        print('Login into mobile')
        
    def security(self):
        print('Mobile security')
```
