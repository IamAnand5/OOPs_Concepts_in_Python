```python
# Fraction Class: Creating your own data type using OOP in Python
# this class is a combination of magic methods and non-magic methods

class Fraction:
    
    # parameterized constructor (which expect/take some inputs while creating object)
    def __init__(self,x,y):
        self.num = x
        self.den = y
        
     # 2nd magic method __str__()   
    def __str__(self):
        return '{}/{}'.format(self.num, self.den) 

    # 3rd magic method (__add__)
    def __add__(self, other):
        new_num = self.num*other.den + other.num*self.den
        new_den = self.den*other.den
        return '{}/{}'.format(new_num, new_den)
    
    # 4th magic method
    def __minus__(self, other):
        new_num = self.num*other.den - other.num*self.den
        new_den = self.den*other.den
        return '{}/{}'.format(new_num, new_den)
    
    #5th magic method
    def __product__(self, other):
        new_num = self.num*other.num
        new_den = self.den*other.den
        return '{}/{}'.format(new_num, new_den)
    
    #6th magic method
    def __truediv__(self, other):
        new_num = self.num*other.den
        new_den = self.den*other.num
        return '{}/{}'.format(new_num, new_den)
    
    # you can also create your own simple methods
    def convert_to_decimal(self):
        return self.num/self.den
```

```python
fr1 = Fraction(5,4)
print(type(fr1))
print(fr1)
fr1.convert_to_decimal()
```

### Pass by reference

```python
class Person:
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender
        
# This is not inside the class. Hence it is a function
def greet(person):   #person is object here
    print('Hi, My name is ',person.name,'I am a ',person.gender)
    p1 = Person('ankit','male')
    return p1    #Here function returning the objects.
    
p = Person('anand', 'male')
x = greet(p)   # p is the object of Person class and refer to person in greet. Now person is object of that class.

print(x.name)
print(x.gender)

# This time we passed an object to a function. And we can do vice versa also. i.e. pass 
# In Python we can pass object as input to the function.
```

```python
# Example Pass by reference
class Person:
    def __init__(self, name, gender):
        self.name = name
        self.gender = gender
        
# Outside the class -> function
def greet(person):   #person is object here
    print(id(person))  #here address will receive
    person.name = 'ankit'
    print(person.name)
    
p = Person('anand', 'male')
print(id(p))   #Technically here we are not passing object to a function. We are passing only address/reference.
greet(p)   
print(p.name)    #now this will print ankit because
```

```python
# Encapsulation
class AtmOperations:
    
    # constructor (special powered function)
    def __init__(self):
        print(id(self))
        self.pin = ''
        self.__balance = 0
        #self.menu()
        
    #creating getter and setter methods to implement the concept of encapsulation
    # The getter()
    def get_balance(self):
        return self.__balance
    
    # The setter()
    def set_balance(self, new_value):
        if type(new_value) == int:
            self.__balance = new_value
        else:
            print('beta bahot marenge')
        
    def menu(self):
        user_input = input('''
        Enter one of the following:
        1. Press 1 to create pin
        2. Press 2 to change pin
        3. Press 3 to check balance
        4. Press 4 to withdraw
        5. Press 5 to exit
        ''')
        
        if user_input == '1':
            #code for create pin
            self.create_pin()
        elif user_input == '2':
            #code for change pin
            self.change_pin()
        elif user_input =='3':
            #code for check balance
            self.check_balance()
        elif user_input == '4':
            # code for withdraw
            self.withdraw()
        else:
            # code for exit
            exit()
        
        
    def create_pin(self):
        user_pin = input('Enter your pin')
        self.pin = user_pin
        
        user_balance = int(input('Enter your balance'))
        self.__balance = user_balance
        
        print('Pin created successfully')
        
        
    def change_pin(self):
        old_pin = input('Enter your old pin')
        
        if old_pin == self.pin:
        # let user to change the pin
            new_pin = input('Enter new pin')
            self.pin = new_pin
            print('Pin changed successfully')
        else:
            print("Old pin not matched")
        
    def check_balance(self):
        user_pin = input('Enter your pin')
        if user_pin == self.pin:
            print("Your balance is ",self.__balance)
        else:
            print("chal nikal yhan se")
            
    def withdraw(self):
        user_pin = input('Enter the pin')
        if user_pin == self.pin:
            #allow user to withdraw
            amount = int(input('Enter the amount'))
            if amount <= self.__balance:
                self.__balance = self.__balance - amount
                print('Withdrawl successful. Balance is', self.__balance)
            else:
                print('Abey Garib')
        else:
            print('Sale Chor')
```

### Difference between Instance variable and Static variable
###### 1.)  Instance variable's value for every object is different; whereas Static variable's value for every object is same.
###### 2.) Instance variable is of  object and Static variable is of class.
###### 3.) Instance variable is defined inside the constructor; whereas Static variable is defined inside the class i.e. outside of all methods

Example:- Customer_name should be Instance variable
Customer_balance should be instance variable
Bank's IFSC code should be Static variable
Student's CGPA should be instance variable
University/College name should be Static variable


### How to identify Instance and Static variable
###### If  any variable name is started with 'self' then it is known as Instance Variable, whereas if a variable name is started with 'class name' then it is called Static variable.

###### Example: Object_name.variable_name = Instance Variable
###### Class_name.variable_name = Static Variable
